### Cell 3 — Reference Performance

Set the reference distance and time per gender that defines "100%" performance.  
These are used to compute reference power via the standard ERG formula `P = 2.8 / pace³`.

```python
import sys
sys.path.insert(0, "..")  # find ergrace package when running from examples/
from ergrace import ERGNormalizer

# ── EDIT THIS CELL ──────────────────────────────────────────────────────────
# Reference performance per gender: distance + time that equals "100%" (score = 1.0).
# Default: World Rowing ergometer records for age 19–29.
#   Men:   5:40 for 2000m  →  340 s
#   Women: 6:40 for 2000m  →  400 s
# Reference: https://worldrowing.com/2026/02/28/records-fall-and-champions-crowned-during-2026-world-rowing-virtual-indoor-championships-presented-by-concept2/

REF_DISTANCE_M = 5000        # reference distance in meters

REF_TIME_MEN_S   = 15*60 + 39  # men's reference time in seconds   (5:40)
REF_TIME_WOMEN_S = 18*60 + 21  # women's reference time in seconds  (6:40)

REF_POWER_MEN   = ERGNormalizer.distance_time2power(REF_DISTANCE_M, REF_TIME_MEN_S)
REF_POWER_WOMEN = ERGNormalizer.distance_time2power(REF_DISTANCE_M, REF_TIME_WOMEN_S)

print(f"Reference power — Men: {REF_POWER_MEN:.1f} W  |  Women: {REF_POWER_WOMEN:.1f} W")
# ────────────────────────────────────────────────────────────────────────────
```

## Results

Run after filling in distances above.

```python
normalizer = ERGNormalizer(
    duration_s=RACE_DURATION_S,
    ref_power_men=REF_POWER_MEN,
    ref_power_women=REF_POWER_WOMEN,
)

for team in TEAMS:
    lane = team["lane"]
    distance = DISTANCES.get(lane, 0)
    if distance == 0:
        continue  # skip teams without a result yet
    label = f"Lane {lane} – {team['club']}"
    normalizer.add_team(label, n_men=team["n_men"], n_women=team["n_women"], distance_20min_m=distance)

if not normalizer.teams:
    print("No distances entered yet. Fill in DISTANCES above and re-run.")
else:
    normalizer.calculate_scores()
    normalizer.print_results()
    normalizer.plot_results()
```
